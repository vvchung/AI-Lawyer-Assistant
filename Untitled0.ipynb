```python
from mempalace.encoders import AAACEncoder

# 初始化轉碼器
encoder = AAACEncoder(language='zh-TW')

palace_data = []

DATA_PATH = '/content/drive/My Drive/Judgment'
files = os.listdir(DATA_PATH)

for filename in files:
    with open(os.path.join(DATA_PATH, filename), 'r', encoding='utf-8') as f:
        content = f.read()

        # 執行 AAAC 轉碼
        # 此步驟會將 Markdown 轉化為高密度的代碼塊
        encoded_block = encoder.encode(
            text=content,
            metadata={"source": filename, "type": "Extraordinary Appeal"}
        )
        palace_data.append(encoded_block)

# 導出為 Palace JSON 檔
import json
with open('supreme_court_memory.json', 'w', encoding='utf-8') as jf:
    json.dump(palace_data, jf, ensure_ascii=False)
```

```python
import json
from mempalace.miner import get_collection
import os

# Define the same persistent ChromaDB path as used by the 'mine' function
persistent_chroma_path = '/content/mempalace_chroma_db'

def export_palace_to_json(output_filename="supreme_court_palace.json"):
    # 1. Get the ChromaDB collection using the persistent path
    collection = get_collection(palace_path=persistent_chroma_path)

    # 2. Retrieve all data from the database
    all_data = collection.get(include=['metadatas', 'documents'])

    palace_export = []

    # 3. Restructure data into Palace JSON format
    # Check if 'ids' exists and is not empty before iterating
    if 'ids' in all_data and all_data['ids']:
        for i in range(len(all_data['ids'])):
            entry = {
                "id": all_data['ids'][i],
                "text": all_data['documents'][i],
                "metadata": all_data['metadatas'][i]
            }
            palace_export.append(entry)

    # 4. Write to JSON file
    with open(output_filename, "w", encoding="utf-8") as f:
        json.dump(palace_export, f, ensure_ascii=False, indent=4)

    print(f"✅ 成功導出 {len(palace_export)} 筆記憶片段至 {output_filename}")

# Execute the export
export_palace_to_json()
```

### 片段數量分佈統計與視覺化

我們將統計每個原始判決書檔案被切割成了多少個記憶片段，並透過視覺化來了解其分佈。

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 統計每個 source_file 的片段數量
fragment_counts = {}
for fragment in palace_fragments:
    source_file = fragment['metadata'].get('source_file')
    if source_file:
        # 只保留文件名，不保留完整路徑，方便閱讀
        filename = os.path.basename(source_file)
        fragment_counts[filename] = fragment_counts.get(filename, 0) + 1

# 轉換為 DataFrame 方便分析與視覺化
fragment_counts_df = pd.DataFrame(list(fragment_counts.items()), columns=['Source File', 'Fragment Count'])

print("--- 判決書檔案片段數量分佈摘要 ---")
print(fragment_counts_df['Fragment Count'].describe())

# 視覺化片段數量的分佈
fig = plt.figure(figsize=(12, 6))
sns.histplot(fragment_counts_df['Fragment Count'], bins=range(1, fragment_counts_df['Fragment Count'].max() + 2), kde=True)
plt.title('Distribution of Fragment Counts per Source File')
plt.xlabel('Number of Fragments')
plt.ylabel('Number of Source Files')
plt.xticks(range(1, fragment_counts_df['Fragment Count'].max() + 2))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 統計每個原始檔案中包含 '違法' 關鍵字的片段數量
keyword_fragment_counts_per_file = {}

for fragment in found_fragments_by_keyword:
    source_file_full_path = fragment['metadata'].get('source_file')
    if source_file_full_path:
        filename = os.path.basename(source_file_full_path)
        keyword_fragment_counts_per_file[filename] = keyword_fragment_counts_per_file.get(filename, 0) + 1

# 將結果轉換為 DataFrame
keyword_fragment_counts_df = pd.DataFrame(list(keyword_fragment_counts_per_file.items()),
                                            columns=['Source File', 'Keyword Fragment Count'])

# 按照數量降序排列，方便查看最多和最少的檔案
keyword_fragment_counts_df = keyword_fragment_counts_df.sort_values(by='Keyword Fragment Count', ascending=False).reset_index(drop=True)

print("--- 包含 '違法' 關鍵字的片段數量分佈 (按檔案) ---")
display(keyword_fragment_counts_df.head(10))

# 視覺化分佈
plt.figure(figsize=(15, 7))
sns.histplot(keyword_fragment_counts_df['Keyword Fragment Count'], bins=range(1, keyword_fragment_counts_df['Keyword Fragment Count'].max() + 2), kde=True)
plt.title("Distribution of '違法' Keyword Fragment Counts per Source File")
plt.xlabel("Number of Fragments Containing '違法'")
plt.ylabel('Number of Source Files')
plt.xticks(range(1, keyword_fragment_counts_df['Keyword Fragment Count'].max() + 2))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
